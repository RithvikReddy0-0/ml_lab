```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from statistics import mean, variance, mode
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from scipy.stats import zscore
from numpy.linalg import norm

excel_path = "Lab_Session_Data.xlsx"

def load_purchase_data():
    return pd.read_excel(excel_path, sheet_name="Purchase data", usecols="A:E")

def get_A_C(df):
    A = df[['Candies (#)','Mangoes (Kg)','Milk Packets (#)']].values
    C = df['Payment (Rs)'].values
    return A, C

def compute_vector_info(A):
    return A.shape[1],A.shape[0],np.linalg.matrix_rank(A),np.linalg.pinv(A)@C

def categorize(df):
    df = df.copy()
    df['Category']=np.where(df['Payment (Rs)']>200,'RICH','POOR')
    return df

def train_model(X, y):
    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
    model = RandomForestClassifier()
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    return classification_report(y_test, y_pred, output_dict=True)

def load_stock_data():
    return pd.read_excel(excel_path, sheet_name="IRCTC Stock Price")

def stock_stats(prices):
    return mean(prices), variance(prices)

def stock_add_features(df):
    df = df.copy()
    df['Date'] = pd.to_datetime(df['Date'])
    df['Weekday'] = df['Date'].dt.day_name()
    df['Month'] = df['Date'].dt.month
    return df

def load_thyroid():
    return pd.read_excel(excel_path, sheet_name="thyroid0387_UCI")

def explore(df):
    res = {}
    for col in df.columns:
        x = df[col]
        res[col] = {
            'type': x.dtype,
            'unique': x.unique()[:5],
            'missing': x.isna().sum(),
            'range': (x.min(), x.max()) if pd.api.types.is_numeric_dtype(x) else None,
            'mean': x.mean() if pd.api.types.is_numeric_dtype(x) else None,
            'variance': x.var() if pd.api.types.is_numeric_dtype(x) else None,
            'outliers': sum(np.abs(zscore(x.dropna())) > 3) if pd.api.types.is_numeric_dtype(x) else None
        }
    return res

def suggest_encoding(df):
    label = []
    onehot = []
    for col in df.columns:
        if df[col].dtype == 'object':
            if df[col].nunique() < 10:
                label.append(col)
            else:
                onehot.append(col)
    return label, onehot

def encode(df, label, onehot):
    df = df.copy()
    for col in label:
        le = LabelEncoder()
        df[col] = le.fit_transform(df[col].astype(str))
    df = pd.get_dummies(df, columns=onehot)
    return df

def jc_smc(v1, v2):
    f11 = np.sum((v1 == 1) & (v2 == 1))
    f00 = np.sum((v1 == 0) & (v2 == 0))
    f10 = np.sum((v1 == 1) & (v2 == 0))
    f01 = np.sum((v1 == 0) & (v2 == 1))
    jc = f11 / (f11 + f10 + f01) if (f11 + f10 + f01) else 0
    smc = (f11 + f00) / (f11 + f10 + f01 + f00) if (f11 + f10 + f01 + f00) else 0
    return jc, smc

def cosine(v1, v2):
    return np.dot(v1, v2) / (norm(v1) * norm(v2)) if norm(v1) * norm(v2) != 0 else 0

def sim_matrices(mat):
    n = mat.shape[0]
    jc = np.zeros((n, n))
    smc = np.zeros((n, n))
    cos = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            jc[i, j], smc[i, j] = jc_smc(mat[i], mat[j])
            cos[i, j] = cosine(mat[i], mat[j])
    return jc, smc, cos

def plot_heat(jc, smc, cos):
    plt.figure(figsize=(15, 5))
    sns.heatmap(jc, ax=plt.subplot(1,3,1)).set(title="JC")
    sns.heatmap(smc, ax=plt.subplot(1,3,2)).set(title="SMC")
    sns.heatmap(cos, ax=plt.subplot(1,3,3)).set(title="Cosine")
    plt.tight_layout()
    plt.show()

def impute(df):
    df = df.copy()
    for col in df.columns:
        if df[col].isna().sum():
            if pd.api.types.is_numeric_dtype(df[col]):
                z = np.abs(zscore(df[col].dropna()))
                fill_value = df[col].mean() if sum(z > 3) == 0 else df[col].median()
                df[col] = df[col].fillna(fill_value)
            else:
                df[col].fillna(mode(df[col].dropna()), inplace=True)
    return df

def normalize(df):
    df = df.copy()
    num_cols = df.select_dtypes(include=np.number).columns
    df[num_cols] = MinMaxScaler().fit_transform(df[num_cols])
    return df


## THe below def are for the optinal section code
def solve_with_square_matrix(A_full, C_full, start_row):
  A_sq = A_full[start_row : start_row + 3]
  C_sq = C_full[start_row : start_row + 3]
  if np.linalg.matrix_rank(A_sq) < 3:
        return f"Matrix starting at row {start_row} is singular. Cannot solve."
  costs = np.linalg.inv(A_sq) @ C_sq
  return costs

if __name__ == "__main__":
    # A1
    df = load_purchase_data()
    A, C = get_A_C(df)
    dim, num_vec, rank, costs = compute_vector_info(A)
    print("A1\n", f"Dim: {dim}, Rows: {num_vec}, Rank: {rank}\nCosts: {costs}")

    # A2
    print('\n')
    df_cat = categorize(df)
    X, y = df_cat[['Candies (#)', 'Mangoes (Kg)', 'Milk Packets (#)']], df_cat['Category']
    print("A2\n", train_model(X, y), df_cat)

    # A3
    print('\n')
    stock = load_stock_data()
    stock = stock_add_features(stock)
    p = stock['Price'].dropna()
    mean_p, var_p = stock_stats(p)
    wed_mean = mean(stock[stock['Weekday'] == 'Wednesday']['Price'])
    apr_mean = mean(stock[stock['Month'] == 4]['Price'])
    prob_loss = (stock['Chg%'] < 0).sum() / len(stock)
    prob_profit_wed = (stock[stock['Weekday'] == 'Wednesday']['Chg%'] > 0).sum() / len(stock[stock['Weekday'] == 'Wednesday'])
    print("A3\n", f"Mean: {mean_p}, Var: {var_p}, WedMean: {wed_mean}, AprMean: {apr_mean}, LossP: {prob_loss}, ProfitOnWed: {prob_profit_wed}")
    sns.stripplot(x=stock['Weekday'], y=stock['Chg%'])
    plt.title("Change % by Day")
    plt.show()

    # A4
    print('\n')
    thyroid = load_thyroid()
    thyroid.replace('?', np.nan, inplace=True)
    for col in ['age', 'TSH', 'T3', 'TT4', 'T4U', 'FTI', 'TBG']:
        thyroid[col] = pd.to_numeric(thyroid[col], errors='coerce')
    info = explore(thyroid)
    labels, onehots = suggest_encoding(thyroid)
    print("A4\n", info)
    print(f"Label: {labels}\nOneHot: {onehots}")

    # A5
    print('\n')
    bin_cols = [c for c in thyroid.columns if set(thyroid[c].dropna().unique()) <= {'t', 'f', 1, 0, 'M', 'F'}]
    thyroid[bin_cols] = thyroid[bin_cols].replace({'t': 1, 'f': 0, 'M': 1, 'F': 0})
    v1 = thyroid.iloc[0][bin_cols].fillna(0).astype(int).values
    v2 = thyroid.iloc[1][bin_cols].fillna(0).astype(int).values
    jc, smc = jc_smc(v1, v2)
    print("A5\n", f"JC: {jc}, SMC: {smc}")

    # A6
    print('\n')
    v1 = pd.to_numeric(thyroid.iloc[0], errors='coerce').fillna(0).values
    v2 = pd.to_numeric(thyroid.iloc[1], errors='coerce').fillna(0).values
    print("A6\n", f"Cosine: {cosine(v1, v2)}")

    # A7
    print('\n')
    bin_mat = thyroid[bin_cols].iloc[:20].fillna(0).astype(int).values
    jc_mat, smc_mat, cos_mat = sim_matrices(bin_mat)
    print("A7")
    plot_heat(jc_mat, smc_mat, cos_mat)

    # A8
    print('\n')
    thyroid_imp = impute(thyroid)
    print("A8\nMissing values after imputation:", thyroid_imp.isnull().sum().sum())
    print(thyroid_imp.head())

    # A9
    print('\n')
    thyroid_norm = normalize(thyroid_imp)
    print("A9\n", thyroid_norm.head())



    print('\n\n')

    print("THe Optional Section")
    print('\n')

    print('----------------Q1-----------\n')
```

```python
print("-----------Optional Section Code-------------\n")

import pandas as pd
import numpy as np
from scipy.stats import zscore
from numpy.linalg import norm
from sklearn.preprocessing import MinMaxScaler

def impute(df):
    x = df.copy()
    for c in x.select_dtypes(include=np.number).columns:
        if x[c].isnull().sum() > 0:
            zs = np.abs(zscore(x[c].dropna()))
            v = x[c].median() if (zs > 3).any() else x[c].mean()
            x[c] = x[c].fillna(v)
    for c in x.select_dtypes(include='object').columns:
        if x[c].isnull().sum() > 0:
            x[c] = x[c].fillna(x[c].mode()[0])
    return x

def normalize(df):
    x = df.copy()
    n = x.select_dtypes(include=np.number).columns
    if len(n): x[n] = MinMaxScaler().fit_transform(x[n])
    return x

def jc_smc(v1, v2):
    a = np.sum((v1 == 1) & (v2 == 1)); b = np.sum((v1 == 0) & (v2 == 0))
    c = np.sum((v1 == 1) & (v2 == 0)); d = np.sum((v1 == 0) & (v2 == 1))
    jc = a / (a + c + d) if (a + c + d) else 0
    smc = (a + b) / (a + b + c + d) if (a + b + c + d) else 0
    return jc, smc

def cosine(v1, v2):
    p = np.dot(v1, v2); d = norm(v1) * norm(v2)
    return p / d if d else 0

try:
    f = "Lab_Session_Data.xlsx"

    print("------ O1 ------")
    d = {
        'Customer': [f'C_{i}' for i in range(1, 11)],
        'Candies (#)': [20, 16, 27, 19, 24, 22, 15, 18, 21, 16],
        'Mangoes (Kg)': [6, 3, 6, 1, 4, 1, 4, 4, 1, 2],
        'Milk Packets (#)': [2, 6, 2, 2, 2, 5, 2, 2, 4, 4],
        'Payment (Rs)': [386, 289, 393, 110, 280, 167, 271, 274, 148, 198]
    }
    df = pd.DataFrame(d)
    A = df.iloc[:, 1:-1].values
    C = df.iloc[:, -1].values

    x_all = np.linalg.pinv(A) @ C

    def solve(A, C, i, n=3):
        if i + n > A.shape[0]: return "Not enough rows"
        try: return np.linalg.inv(A[i:i+n]) @ C[i:i+n]
        except: return "Singular"

    x1 = solve(A, C, 0)
    x2 = solve(A, C, 3)

    print(f"x full: {np.round(x_all, 2)}")
    print(f"x 0-2: {np.round(x1, 2) if isinstance(x1, np.ndarray) else x1}")
    print(f"x 3-5: {np.round(x2, 2) if isinstance(x2, np.ndarray) else x2}")
    print("→ small subset unstable; full pinv best fit")

    print("\n\n------ O2 ------")
    d_raw = pd.read_excel(f, sheet_name="thyroid0387_UCI").replace('?', np.nan)
    for c in ['age', 'TSH', 'T3', 'TT4', 'T4U', 'FTI', 'TBG']:
        d_raw[c] = pd.to_numeric(d_raw[c], errors='coerce')
    full = impute(d_raw)
    samp = full.sample(20, random_state=42).reset_index(drop=True)

    b = [c for c in samp.columns if samp[c].dropna().isin(['t', 'f']).all()]
    bin_df = samp[b].replace({'t': 1, 'f': 0})
    jc, smc = jc_smc(bin_df.iloc[0].values, bin_df.iloc[1].values)
    print(f"jc: {jc:.4f}, smc: {smc:.4f}")

    v1 = samp.select_dtypes(include=np.number).iloc[0].values
    v2 = samp.select_dtypes(include=np.number).iloc[1].values
    print(f"cos: {cosine(v1, v2):.4f}")

    print("\n\n------ O3 ------")
    m = pd.read_excel(f, sheet_name="marketing_campaign")
    m_imp = impute(m)
    m_norm = normalize(m_imp)
    print(f"shape: {m.shape}, nulls: {m_imp.isnull().sum().sum()}")
    print(m_norm.head())

except FileNotFoundError:
    print(f"{f} not found.")
except Exception as e:
    print(f"err: {e}")
```
